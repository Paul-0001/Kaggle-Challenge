Feature engineering: 
This notebook documents a really broad set of hypotheses, not features already shown to improve predictive performance.The real use of these features will be determined later on modelspecifically. The original columns are preserved. No feature uses the buy label because then we would use the predicted value to predict. Missing values are intentionally preserved (See EDA to see why it is important) for undefined ratios and absent listening histories. Genre similarities (Used for our genre "Distance matrix") are inferred from unlabeled app-user listening profiles and therefore measure co-listening audiences, not objective music theory (Because even if the theory might be more true in reality in our data set it might not be which is yh we do it empirically). The estimated total price assumes the percentage discount applies only to the ticket price not to the booking fee as well. All timestamps are UTC, and music-app listening events on or after the prediction cutoff are excluded (else data leakage).

Names follow one consistent convention: *_squared, log_*, *_ratio, *_difference, and original_feature_x_other_feature. Names for derived feature interactions reuse the full derived name. For categorical interactions, age is bucketed into decades to avoid high-cardinality combinations.

Build train and test features:

```python

# Import what needed
from pathlib import Path
import sys
import pandas as pd

# Set project root
PROJECT_ROOT = Path.cwd().resolve()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

# Insert scripts into Python import path
sys.path.insert(0, str(PROJECT_ROOT))

# Import our functions
from scripts.feature_engineering import build_features, check_feature_integrity
from scripts.split_data import split_by_user

# Find raw data
raw = PROJECT_ROOT / "data" / "raw"

# Read train and test datasets
train = pd.read_csv(raw / "train.csv")
test = pd.read_csv(raw / "test.csv")

# Determine training users BEFORE feature engineering
train_part, holdout_part = split_by_user(train, test_size=0.20, random_state=42)

# Only these users may be used to learn genre similarity
similarity_user_ids = set(train_part["music_app_user_id"].dropna())

# Global safe cutoff across train and test
safe_cutoff = min(pd.to_datetime(train["prediction_at"], utc=True).min(),pd.to_datetime(test["prediction_at"], utc=True).min())

# Build train features
train_features = build_features(train, raw / "music_app.json",safe_cutoff=safe_cutoff, similarity_user_ids=similarity_user_ids)

# Build test features
test_features = build_features(test, raw / "music_app.json",safe_cutoff=safe_cutoff, similarity_user_ids=similarity_user_ids)

# Check feature integrity
check_feature_integrity(train_features, test_features)

# Display results
print("Train:", train_features.shape)
print("Test:", test_features.shape)
print("Added features:", len(train_features.columns) - len(train.columns))
display(train_features.head())

```

```python
# Set output path in processed data subfolder
out = PROJECT_ROOT / "data" / "processed"

# Create parent folders if they do not exist (They should)
out.mkdir(parents=True, exist_ok=True)

# Save the new train dataset with new features as CSV
train_features.to_csv(out / "train_features.csv", index=False)

# Save new test dataset with new features as CSV
test_features.to_csv(out / "test_features.csv", index=False)

# Usercheck to see if path of download is correct
print('Feature datasets saved to', out)
```
